```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks  # dictionary {subject: score}

    def validate_marks(self):
        for subject, score in self.marks.items():
            if not isinstance(score, (int, float)):
                raise ValueError(f"Invalid mark for {subject}: must be a number")
            if score < 0 or score > 100:
                raise ValueError(f"Invalid mark for {subject}: must be between 0 and 100")

    def total_marks(self):
        return sum(self.marks.values())

    def average_marks(self):
        if len(self.marks) == 0:
            raise ValueError("No subjects provided")
        return self.total_marks() / len(self.marks)

    def grade(self):
        avg = self.average_marks()
        if avg >= 90:
            return "A+"
        elif avg >= 75:
            return "A"
        elif avg >= 60:
            return "B"
        elif avg >= 40:
            return "C"
        else:
            return "F"

    def pass_fail(self):
        return "Pass" if all(score >= 40 for score in self.marks.values()) else "Fail"

    def result_summary(self):
        self.validate_marks()
        print(f"--- Result for {self.name} (Roll No: {self.roll_no}) ---")
        for subject, score in self.marks.items():
            print(f"{subject}: {score}")
        print(f"Total Marks: {self.total_marks()}")
        print(f"Average Marks: {self.average_marks():.2f}")
        print(f"Grade: {self.grade()}")
        print(f"Result: {self.pass_fail()}")
        print("-------------------------------------------")


# Example usage
if __name__ == "__main__":
    marks = {
        "Math": 85,
        "Science": 78,
        "English": 92,
        "History": 70,
        "Computer": 88
    }

    student1 = Student("Alice", 101, marks)
    student1.result_summary()
```

Edge Cases to Handle
Missing data

Student name or roll number not provided.

Marks missing for one or more subjects.

Invalid marks

Marks entered as text instead of numbers.

Marks outside valid range (e.g., negative values or >100).

Empty subject list

No subjects provided → division by zero when calculating average.

Single subject case

Only one subject → average equals that subject’s score.

Fail condition

Student fails if any subject < 40, even if average is high.

Boundary values

Marks exactly at grade thresholds (e.g., 90 → A+, 75 → A).

Marks exactly at pass threshold (40 → Pass).

Multiple students

Analyzer should handle a batch of students without crashing.

Duplicate roll numbers

Two students with the same roll number → need uniqueness check.

Large dataset

Hundreds of students → performance considerations.

Special cases

Student with perfect scores (100 in all subjects).

Student with zero in all subjects.
